- `regla_30` (>30 °C, <30 % de humedad y >30 km/h);
- `antiguedad_min`: minutos desde la medida. No se descarta ninguna estación por ir atrasada.

Solo produce datos, no abre ningún mapa: para verlos en el mapa interactivo está el pipeline completo (`pipeline_incendios.ipynb`).

## Revisión de librerías

Ejecuta estas celdas antes que nada. Este reto necesita: crawl4ai (descarga), numpy y pandas.
Si falta alguna, la celda lo dice y explica cómo instalarla.

```python
# Revisión de librerías: ejecútala PRIMERO.
# Comprueba que cada librería está instalada en ESTE kernel (y con la versión que pide requirements.txt).
import importlib.metadata as md
import re
import sys
from pathlib import Path

# Mínimas de este reto; si hay requirements.txt junto al notebook, mandan sus versiones.
NECESARIAS = {'crawl4ai': '', 'numpy': '', 'pandas': ''}

pedidas = dict(NECESARIAS)
req = Path("requirements.txt")
if req.exists():
    for linea in req.read_text(encoding="utf-8").splitlines():
        m = re.match(r"\s*([A-Za-z0-9_.\-]+)\s*((?:[=<>!~]=?|===)\s*[^\s#;]+)?", linea.split("#")[0])
        if m and m.group(1).lower() in {k.lower() for k in pedidas}:
            pedidas[next(k for k in pedidas if k.lower() == m.group(1).lower())] = (m.group(2) or "").replace(" ", "")

try:
    from packaging.specifiers import SpecifierSet
    from packaging.version import Version
except ImportError:          # sin `packaging` solo se comprueba que estén instaladas
    SpecifierSet = Version = None

filas, faltan, versiones = [], [], []
for paquete, requisito in pedidas.items():
    try:
        instalada = md.version(paquete)
    except md.PackageNotFoundError:
        filas.append(f"  ✗ {paquete:<10} no instalada" + (f"   (pide {requisito})" if requisito else ""))
        faltan.append(paquete)
        continue
    ok = True
    if requisito and SpecifierSet:
        ok = Version(instalada) in SpecifierSet(requisito if requisito[0] in "<>=!~" else "==" + requisito)
    filas.append(f"  {'✓' if ok else '!'} {paquete:<10} {instalada}" + ("" if ok else f"   (pide {requisito})"))
    if not ok:
        versiones.append(paquete)

print(f"Python {sys.version.split()[0]} · {sys.executable}\n")
print("\n".join(filas))
if faltan:
    raise ImportError(
        "Faltan librerías: " + ", ".join(faltan) + ".\n"
        "Instálalas en este kernel con la celda siguiente (descomenta la línea), reinicia el kernel y vuelve a ejecutar:\n"
        "    %pip install -r requirements.txt pandas")
if versiones:
    print("\nAviso: hay versiones distintas a las de requirements.txt (" + ", ".join(versiones) + "). Puede funcionar, pero no está probado.")
else:
    print("\nTodo en orden ✓")
```

### 2.2 Condiciones de riesgo

Estaciones con la humedad más baja y las que cumplen la regla del 30, el umbral de condiciones extremas de propagación.

```python
print("Estaciones en regla del 30:", int(df["regla_30"].fillna(False).sum()))
display(df.sort_values("humedad_pct")[columnas].head(10))
```

### 2.3 Antigüedad de las lecturas

No se descarta ninguna estación por ir atrasada; aquí se ve cuántas tienen una lectura reciente.

```python
tramos = pd.cut(df["antiguedad_min"], [-1, 30, 90, 180, 360, float("inf")],
                labels=["≤ 30 min", "30–90 min", "1,5–3 h", "3–6 h", "> 6 h"])
display(tramos.value_counts().sort_index().rename("estaciones").to_frame())
```
